```python
import numpy as np 
import matplotlib.pyplot as plt 
 
x = np.array([1, 2, 3], dtype=float) 
y = np.array([2, 3, 4], dtype=float) 
 
n = len(x) 
 
 
w = 0.0 
b = 0.0 
alpha = 0.1 
iterations = 3 
 
print("Input x :", x) 
print("Actual y :", y) 
print("Number of samples :", n) 
print("Initial weight (w) :", w) 
print("Initial bias (b) :", b) 
print("Learning rate :", alpha) 
 
 
y_pred = w * x + b 
cost = np.sum((y - y_pred) ** 2) / n 
 
print("Predictions :", y_pred) 
print("\nInitial prediction :", y_pred) 
print(f"Initial Cost : {cost:.4f}") 
 
costs = [] 
w_history = [] 
b_history = [] 
 
 
for i in range(iterations): 
 
 
    y_pred = w * x + b 
 
 
    dw = np.sum(-2 * x * (y - y_pred)) / n 
    db = np.sum(-2 * (y - y_pred)) / n 
 
 
    w = w - alpha * dw 
    b = b - alpha * db 
 
 
    y_pred = w * x + b 
    cost = np.sum((y - y_pred) ** 2) / n 
 
    costs.append(cost) 
    w_history.append(w) 
    b_history.append(b) 
 
    print("\n" + "-" * 30) 
    print(f"Iteration {i + 1}") 
    print(f"The y_predict (after update) is {np.round(y_pred, 4)}") 
    print(f"The cost (after update) is {cost:.4f}") 
    print(f"Gradients dw: {dw:.4f}, db: {db:.4f}") 
    print(f"The updated weight (w) is {w:.4f}") 
    print(f"The updated bias (b) is {b:.4f}") 
 
 
 
print("\nThe final equation is :") 
print(f"y = {w:.4f}x + {b:.4f}") 
print("The final cost is", costs[-1]) 
 
 
 
plt.plot(range(1, iterations + 1), costs, marker='o') 
plt.xlabel("Iterations") 
plt.ylabel("Cost") 
plt.title("Cost Function VS Iterations") 
plt.show() 
 
 
plt.scatter(x, y, label="Actual Data") 
 
final_y = w * x + b 
plt.plot(x, final_y, label="Regression Line") 
 
plt.xlabel("x") 
plt.ylabel("y") 
plt.title("Linear Regression Fit") 
plt.legend() 
plt.show()  
```
